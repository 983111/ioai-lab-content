# Text Classification Baseline

Build a TF-IDF + logistic regression classifier. A small toy dataset keeps this notebook self-contained and fast.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,ConfusionMatrixDisplay,accuracy_score
texts=['win a free prize now','claim your cash reward','limited offer click now','you have won a gift','urgent claim your bonus','free contest entry','exclusive deal buy today','get money fast','team meeting moved to monday','send the project report','lunch at noon today','lecture starts at ten','review my assignment','see you at the library','meeting notes are attached','happy birthday to you','parcel arrives tomorrow','thanks for your help','call me after class','exam timetable is ready']
labels=['spam']*8+['ham']*12
df=pd.DataFrame({'text':texts,'label':labels})
display(df.head())
```

## 1. Inspect class balance

```python
print(df.label.value_counts())
print('Rows:',len(df),'Missing:',df.isna().sum().to_dict())
```

## 2. Split before fitting vectorizer

```python
X_train,X_test,y_train,y_test=train_test_split(df.text,df.label,test_size=.30,random_state=42,stratify=df.label)
print(len(X_train),len(X_test))
```

## 3. Train a leakage-safe pipeline

TF-IDF learns its vocabulary from training text only because it is inside the pipeline.

```python
model=make_pipeline(TfidfVectorizer(ngram_range=(1,2),stop_words='english'),LogisticRegression(max_iter=1000,class_weight='balanced',random_state=42))
model.fit(X_train,y_train)
pred=model.predict(X_test)
print('Accuracy:',accuracy_score(y_test,pred))
print(classification_report(y_test,pred,zero_division=0))
```

## 4. Confusion matrix

```python
ConfusionMatrixDisplay.from_predictions(y_test,pred,labels=['ham','spam'],display_labels=['ham','spam']); plt.show()
```

## 5. Try new messages

These outputs demonstrate the workflow only; the tiny toy dataset is not suitable for production.

```python
messages=['claim your free reward today','meet me after the lecture','your report is ready']
for text,label in zip(messages,model.predict(messages)): print(label,'|',text)
```
