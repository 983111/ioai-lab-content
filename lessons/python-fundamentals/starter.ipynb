## 2. Conditions and loops

Count how many scores are at least 75. Try changing the threshold and the input values.

```python
scores = [72, 85, 91, 68, 77, 75]
threshold = 75
passed = 0

for score in scores:
    if score >= threshold:
        passed += 1

print("Number meeting threshold:", passed)
```

## 3. Collections

Use a dictionary to count the number of times each label occurs.

```python
labels = ["healthy", "diseased", "healthy", "healthy", "diseased"]
counts = {}

for label in labels:
    counts[label] = counts.get(label, 0) + 1

print(counts)
```

## 4. Write a reusable function

Complete `calculate_mean`. The function should reject an empty list and return the arithmetic mean.

```python
def calculate_mean(values):
    if len(values) == 0:
        raise ValueError("values must not be empty")
    # TODO: calculate and return the mean
    return sum(values) / len(values)

measurements = [28.5, 30.0, 29.5, 31.0, 27.0]
print("Mean measurement:", calculate_mean(measurements))
```

## 5. Mini-project: dataset summary

Complete the function below. It must return the count, mean, minimum, maximum, and number of values at least equal to the threshold.

```python
def summarize_scores(scores, threshold=75):
    if not scores:
        raise ValueError("scores must not be empty")

    # TODO: return a dictionary with the required summary fields
    return {
        "count": len(scores),
        "mean": sum(scores) / len(scores),
        "minimum": min(scores),
        "maximum": max(scores),
        "at_least_threshold": sum(1 for score in scores if score >= threshold)
    }

sample_scores = [72, 85, 91, 68, 77]
report = summarize_scores(sample_scores)

for key, value in report.items():
    print(f"{key}: {value}")
```
